```python
#recentMatches_info_table

RM_rows = []

for info in RM_matchInfo:

    start_ts = info.get('startDate')
    end_ts = info.get('endDate')

   #start_date = datetime.fromtimestamp(int(start_ts) / 1000).strftime('%d/%m/%Y') if start_ts else None
    #end_date = datetime.fromtimestamp(int(end_ts) / 1000).strftime('%d/%m/%Y') if end_ts else None

    start_date = datetime.fromtimestamp(int(start_ts) / 1000) if start_ts else None
    end_date = datetime.fromtimestamp(int(end_ts) / 1000) if end_ts else None

    row = {
        "matchId": info.get('matchId'),
        "seriesId": info.get('seriesId'),
        "seriesName": info.get('seriesName'),
        "matchDesc": info.get('matchDesc'),
        "startDate": start_date,
        "endDate": end_date,
        "team1": info['team1']['teamSName'],
        "team2": info['team2']['teamSName'],
        "venue" : info['venueInfo']['ground'],
        "city": info['venueInfo']['city'],
        "status": info.get('status')
    }

    RM_rows.append(row)

RM = pd.DataFrame(RM_rows)
print(RM)
```

```python


rr_player_ids = set()
for match in rr_matches_sc:
    for innings in match['scorecard']:
        if innings.get('batteamname') == 'Rajasthan Royals':
            for batsman in innings['batsman']:
                rr_player_ids.add(batsman['id'])

# Initialize player stats dictionary for RR players only
player_stats = {}

# Process each match
for match in rr_matches_sc:
    for innings in match['scorecard']:
        # Process RR batting innings
        if innings.get('batteamname') == 'Rajasthan Royals':
            for batsman in innings['batsman']:
                player_id = batsman['id']
                player_name = batsman['name']
                
                if player_id not in player_stats:
                    player_stats[player_id] = {
                        'player_id': player_id,
                        'player_name': player_name,
                        'matches': 0,
                        'innings_batted': 0,
                        'runs': 0,
                        'balls_faced': 0,
                        'not_outs': 0,
                        'wickets': 0,
                        'balls_bowled': 0,
                        'runs_conceded': 0,
                        'catches': 0,
                        'stumpings': 0,
                        'runouts': 0
                    }
                
                # Increment matches
                player_stats[player_id]['matches'] += 1
                
                # Update batting stats
                if batsman['balls'] > 0:
                    player_stats[player_id]['innings_batted'] += 1
                    player_stats[player_id]['runs'] += batsman['runs']
                    player_stats[player_id]['balls_faced'] += batsman['balls']
                    
                    # Check if not out
                    if 'not out' in batsman.get('outdec', '').lower():
                        player_stats[player_id]['not_outs'] += 1
        
        # Process RR bowling innings (when opponent is batting)
        else:
            # Check if 'bowler' key exists in innings
            if 'bowler' in innings:
                for bowler in innings['bowler']:
                    player_id = bowler['id']
                    
                    # Only process if this is an RR player
                    if player_id in rr_player_ids:
                        if player_id not in player_stats:
                            player_stats[player_id] = {
                                'player_id': player_id,
                                'player_name': bowler['name'],
                                'matches': 1,
                                'innings_batted': 0,
                                'runs': 0,
                                'balls_faced': 0,
                                'not_outs': 0,
                                'wickets': 0,
                                'balls_bowled': 0,
                                'runs_conceded': 0,
                                'catches': 0,
                                'stumpings': 0
                            }
                        
                        # Update bowling stats
                        player_stats[player_id]['wickets'] += bowler['wickets']
                        player_stats[player_id]['balls_bowled'] += bowler['balls']
                        player_stats[player_id]['runs_conceded'] += bowler['runs']

# Process catches and stumpings from opponent dismissals (outdec)
for match in rr_matches_sc:
    for innings in match['scorecard']:
        # Only check opponent batting innings
        if innings.get('batteamname') != 'Rajasthan Royals':
            for batsman in innings['batsman']:
                outdec = batsman.get('outdec', '')
                
                # Check for catches
                if 'c ' in outdec and ' b ' in outdec:
                    # Extract fielder name from "c FielderName b BowlerName"
                    try:
                        fielder_part = outdec.split('c ')[1].split(' b ')[0].strip()
                        # Match with RR players
                        for pid, pstats in player_stats.items():
                            player_name_lower = pstats['player_name'].lower()
                            fielder_part_lower = fielder_part.lower()
                            # Check various name matches
                            if (fielder_part_lower == player_name_lower or 
                                fielder_part_lower in player_name_lower or 
                                player_name_lower in fielder_part_lower or
                                # Handle specific cases
                                (fielder_part_lower == 'buttler' and 'buttler' in player_name_lower) or
                                (fielder_part_lower == 'jurel' and 'jurel' in player_name_lower) or
                                (fielder_part_lower == 'burger' and 'burger' in player_name_lower)):
                                player_stats[pid]['catches'] += 1
                                break
                    except:
                        pass
                
                # Check for runouts
                if 'run out' in outdec.lower():
                    # Extract fielder name(s) from various runout formats
                    # Examples: "run out (Fielder1)", "run out (Fielder1/Fielder2)", "run out (Fielder)"
                    try:
                        if '(' in outdec and ')' in outdec:
                            fielders_part = outdec.split('(')[1].split(')')[0].strip()
                            # Handle multiple fielders separated by /
                            fielders = [f.strip() for f in fielders_part.split('/')]
                            
                            for fielder in fielders:
                                # Match with RR players
                                for pid, pstats in player_stats.items():
                                    player_name_lower = pstats['player_name'].lower()
                                    fielder_lower = fielder.lower()
                                    if (fielder_lower == player_name_lower or 
                                        fielder_lower in player_name_lower or 
                                        player_name_lower in fielder_lower or
                                        # Handle specific cases
                                        (fielder_lower == 'buttler' and 'buttler' in player_name_lower) or
                                        (fielder_lower == 'jurel' and 'jurel' in player_name_lower) or
                                        (fielder_lower == 'samson' and 'samson' in player_name_lower)):
                                        player_stats[pid]['runouts'] += 1
                                        break
                    except:
                        pass
                
                # Check for stumpings
                if 'st ' in outdec:
                    # Extract keeper name from "st KeeperName b BowlerName"
                    try:
                        keeper_part = outdec.split('st ')[1].split(' b ')[0].strip()
                        # Match with RR players
                        for pid, pstats in player_stats.items():
                            player_name_lower = pstats['player_name'].lower()
                            keeper_part_lower = keeper_part.lower()
                            if (keeper_part_lower == player_name_lower or 
                                keeper_part_lower in player_name_lower or 
                                player_name_lower in keeper_part_lower):
                                player_stats[pid]['stumpings'] += 1
                                break
                    except:
                        pass

# Calculate averages and rates
for player_id, stats in player_stats.items():
    # Ensure runouts field exists
    if 'runouts' not in stats:
        stats['runouts'] = 0
    
    # Batting average
    innings_dismissed = stats['innings_batted'] - stats['not_outs']
    if innings_dismissed > 0:
        stats['batting_avg'] = round(stats['runs'] / innings_dismissed, 2)
    else:
        stats['batting_avg'] = 0 if stats['runs'] == 0 else stats['runs']
    
    # Strike rate
    if stats['balls_faced'] > 0:
        stats['strike_rate'] = round((stats['runs'] / stats['balls_faced']) * 100, 2)
    else:
        stats['strike_rate'] = 0
    
    # Bowling average
    if stats['wickets'] > 0:
        stats['bowling_avg'] = round(stats['runs_conceded'] / stats['wickets'], 2)
    else:
        stats['bowling_avg'] = 0
    
    # Economy rate
    if stats['balls_bowled'] > 0:
        overs = stats['balls_bowled'] / 6
        stats['economy'] = round(stats['runs_conceded'] / overs, 2)
    else:
        stats['economy'] = 0

# Convert to list of dictionaries
rr_players_stats = []
for player_id, stats in sorted(player_stats.items(), key=lambda x: x[1]['runs'], reverse=True):
    rr_players_stats.append({
        'player_id': stats['player_id'],
        'player_name': stats['player_name'],
        'matches': stats['matches'],
        'runs': stats['runs'],
        'balls_faced': stats['balls_faced'],
        'batting_avg': stats['batting_avg'],
        'strike_rate': stats['strike_rate'],
        'wickets': stats['wickets'],
        'bowling_avg': stats['bowling_avg'],
        'economy': stats['economy'],
        'catches': stats['catches'],
        'stumpings': stats['stumpings'],
        'runouts': stats['runouts']
    })
```

Study successful batting partnerships to identify the best player combinations. For pairs of players who have batted together as consecutive batsmen (positions differ by 1) in at least 5 partnerships:
Calculate their average partnership runs
Count how many of their partnerships exceeded 50 runs
Find their highest partnership score
Calculate their success rate (percentage of good partnerships)
Rank the most successful batting partnerships.


```python
partnership_rows = []

for match in total_matches_sc:
    for inng in match["scorecard"]:        
        batting_map = {}
        bats_list = inng.get("batsman", [])
        for i, b in enumerate(bats_list, start=1):
            batting_map[b["id"]] = i  

        
        if "partnership" not in inng:
            continue
        
        plist = inng["partnership"].get("partnership", [])

        for p in plist:
            bat1 = p["bat1id"]
            bat2 = p["bat2id"]
            name1 = p["bat1name"]
            name2 = p["bat2name"]
            runs = p["totalruns"]
            balls = p["totalballs"]

            
            pos1 = batting_map.get(bat1)
            pos2 = batting_map.get(bat2)

            
            if pos1 is None or pos2 is None:
                continue

            if abs(pos1 - pos2) == 1:
                
                if pos1 < pos2:
                    p1_id, p1_name = bat1, name1
                    p2_id, p2_name = bat2, name2
                else:
                    p1_id, p1_name = bat2, name2
                    p2_id, p2_name = bat1, name1

                partnership_rows.append({
                    "player1_id": p1_id,
                    "player1_name": p1_name,
                    "player2_id": p2_id,
                    "player2_name": p2_name,
                    "runs": runs,
                    "balls": balls,
                    "player1_pos": min(pos1, pos2),
                    "player2_pos": max(pos1, pos2)
                })
```

```python
partnership_rows
```

```python

dfp = pd.DataFrame(partnership_rows)

groups = dfp.groupby(["player1_id", "player2_id", "player1_name", "player2_name"])

results = []

for (p1_id, p2_id, p1_name, p2_name), g in groups:
    
    total_ps = len(g)

    
    if total_ps < 5:
        continue

    avg_runs = g["runs"].mean()
    fifty_plus = (g["runs"] >= 50).sum()
    highest = g["runs"].max()

    success_rate = (fifty_plus / total_ps) * 100

    results.append({
        "player1_id": p1_id,
        "player1_name": p1_name,
        "player2_id": p2_id,
        "player2_name": p2_name,
        "total_partnerships": total_ps,
        "avg_runs": round(avg_runs, 2),
        "fifty_plus_partnerships": fifty_plus,
        "highest_partnership": highest,
        "success_rate_percent": round(success_rate, 2)
    })

partner_summary_df = pd.DataFrame(results)
print(partner_summary_df)
```
